```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("ankitbansal06/retail-orders")

print("Path to dataset files:", path)
```

```python
import pandas as pd
#read data from file and handle null values
df = pd.read_csv(r"C:\PyProj\orders.csv",na_values = ['Not Available', 'unknown'])
```

```python
#rename column names 
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
#new discount column, sale price and profit
df['discount']=df['list_price']*df['discount_percent']/100
df['sale_price']=df['list_price']-df['discount']
df['profit']=df['sale_price']-df['cost_price']
```

```python
#convert order date from object to day time 
df['order_date']=pd.to_datetime(df['order_date'],format = "%Y-%m-%d")
df.dtypes
```

```python
#remove cost price ,list price and discount percent columns
df.drop(columns=['cost_price','list_price','discount_percent'],inplace = True)
```
